```python
import numpy as np
import matplotlib.pyplot as plt
```

## Variablendefinition

Startposition

```python
q0 = 0
```

Endposition

```python
q1 = 0.5
```

Maximale Geschwindigkeit

```python
qd_max = 1
```

Beschleunigungen

```python
qdd_acc = 10
qdd_dec = -10
```

Startzeit

```python
t0 = 0
```

## Berechnung der Zeitpunkte $t_1, t_2$ und $t_3$

Fall 1: Maximale Geschwindigkeit $\dot q_{\max}$ wird nicht erreicht

Fall 2: Maximale Geschwindigkeit $\dot q_{\max}$ wird erreicht

```python
# Fall 1
if qd_max > np.sqrt(2*(qdd_acc*qdd_dec)/(qdd_dec - qdd_acc)*abs(q1 - q0)):
    qd_max = np.sqrt(2*(qdd_acc*qdd_dec)/(qdd_dec - qdd_acc)*abs(q1 - q0))
    t1 = qd_max/qdd_acc + t0
    t2 = t1
    t3 = -qd_max/qdd_dec + t2
# Fall 2
else:
    t1 = qd_max/qdd_acc + t0
    t2 = abs(q1 - q0)/qd_max - qd_max/(2*qdd_acc) + qd_max/(2*qdd_dec) + t1
    t3 = -qd_max/qdd_dec + t2
```

## Umkehren der Vorzeichen, falls Bewegung in negative Richtung

```python
if q1 < q0:
    qd_max = -qd_max
    qdd_acc = -qdd_acc
    qdd_dec = -qdd_dec
```

## Zeitdiskrete Berechnung der Bahn

```python
def trange(a, b, step):
    """Hilfsfunktion für die Erzeugung von Zeitvektoren.

    Gibt ein Array mit Werten von a bis b in Schritten von step zurück.
    Der Endpunkt b ist inklusive.
    """
    n = int(round((b - a)/step))
    return a + step*np.arange(n + 1)


dt = 0.01
t = trange(t0, t3, dt)
q = np.zeros(len(t))
q[0] = q0
qd = np.zeros(len(t))

for i in range(len(t) - 1):
    if t[i] < t1:
        qd[i + 1] = qd[i] + qdd_acc*dt
        q[i + 1] = q[i] + qd[i]*dt
    elif t[i] < t2:
        qd[i + 1] = qd[i]
        q[i + 1] = q[i] + qd[i]*dt
    else:
        qd[i + 1] = qd[i] + qdd_dec*dt
        q[i + 1] = q[i] + qd[i]*dt
```

## Plotten

```python
fig, axs = plt.subplots(2, 1, figsize=(8, 7))

axs[0].plot(t, q)
axs[0].set_title("Position")
axs[0].set_xlabel("t")
for tk, label in [(t0, "t0"), (t1, "t1"), (t2, "t2"), (t3, "t3")]:
    axs[0].axvline(tk, linestyle='--', color='gray', linewidth=0.8)
    axs[0].text(tk, axs[0].get_ylim()[1], label, ha='center', va='bottom', fontsize=8)

axs[1].plot(t, qd)
axs[1].set_title("Geschwindigkeit")
axs[1].set_xlabel("t")
for tk, label in [(t0, "t0"), (t1, "t1"), (t2, "t2"), (t3, "t3")]:
    axs[1].axvline(tk, linestyle='--', color='gray', linewidth=0.8)
    axs[1].text(tk, axs[1].get_ylim()[1], label, ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```
