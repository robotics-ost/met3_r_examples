```python
import sympy as sp
from sympy import symbols, cos, sin, Matrix, simplify
from IPython.display import display

sp.init_printing()
```

## Aufstellen der homogenen Transformationsmatrizen

```python
q1, q2, q3, q4, l0, l1, l2, l3, l4 = symbols('q1 q2 q3 q4 l0 l1 l2 l3 l4')

T0_1 = Matrix([
    [cos(q1), -sin(q1), 0, 0],
    [sin(q1),  cos(q1), 0, 0],
    [0,        0,       1, l0],
    [0,        0,       0, 1]])
print("T0_1 = ")
display(T0_1)

T1_2 = Matrix([
    [1, 0,        0,        0],
    [0, cos(q2), -sin(q2),  0],
    [0, sin(q2),  cos(q2),  l1],
    [0, 0,        0,        1]])
print("T1_2 = ")
display(T1_2)

T2_3 = Matrix([
    [1, 0, 0, 0],
    [0, 1, 0, l2 + q3],
    [0, 0, 1, 0],
    [0, 0, 0, 1]])
print("T2_3 = ")
display(T2_3)

T3_4 = Matrix([
    [1, 0,        0,        0],
    [0, cos(q4), -sin(q4),  l3],
    [0, sin(q4),  cos(q4),  0],
    [0, 0,        0,        1]])
print("T3_4 = ")
display(T3_4)

T4_TCP = Matrix([
    [1, 0, 0, 0],
    [0, 1, 0, l4],
    [0, 0, 1, 0],
    [0, 0, 0, 1]])
print("T4_TCP = ")
display(T4_TCP)

T0_TCP = T0_1 * T1_2 * T2_3 * T3_4 * T4_TCP
print("T0_TCP = T0_1 * T1_2 * T2_3 * T3_4 * T4_TCP = ")
display(T0_TCP)
print(" = ")
T0_TCP = simplify(T0_TCP)
display(T0_TCP)
```

## Ermitteln der Vorwärtskinematik

### Herauslesen der Position des Endeffektors

```python
x = T0_TCP[0, 3]
y = T0_TCP[1, 3]
z = T0_TCP[2, 3]

print("x = ")
display(x)
print("y = ")
display(y)
print("z = ")
display(z)
```

### Ermitteln der Orientierung (Z-Y-X Euler Winkel)

```python
phi_x, phi_y, phi_z = symbols('phi_x phi_y phi_z')

Rx = Matrix([
    [1, 0,          0],
    [0, cos(phi_x), -sin(phi_x)],
    [0, sin(phi_x),  cos(phi_x)]])

Ry = Matrix([
    [ cos(phi_y), 0, sin(phi_y)],
    [ 0,          1, 0],
    [-sin(phi_y), 0, cos(phi_y)]])

Rz = Matrix([
    [cos(phi_z), -sin(phi_z), 0],
    [sin(phi_z),  cos(phi_z), 0],
    [0,           0,          1]])

Rzyx = Rz * Ry * Rx
print("Rzyx = Rz * Ry * Rx = ")
display(Rzyx)

phi_y = simplify(sp.asin(-T0_TCP[2, 0]))
print("phi_y = ")
display(phi_y)

phi_x = simplify(sp.atan((T0_TCP[2, 1] / sp.cos(phi_y)) / (T0_TCP[2, 2] / sp.cos(phi_y))))
print("phi_x = ")
display(phi_x)

phi_z = simplify(sp.atan((T0_TCP[1, 0] / sp.cos(phi_y)) / (T0_TCP[0, 0] / sp.cos(phi_y))))
print("phi_z = ")
display(phi_z)
```
