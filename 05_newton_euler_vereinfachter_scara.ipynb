```python
import sympy as sp
from IPython.display import display

sp.init_printing()
```

## Variablendeklaration

```python
l1, l2, m1, m2, I1, I2, lc1, lc2, g = sp.symbols('l1 l2 m1 m2 I1 I2 lc1 lc2 g')

t = sp.symbols('t')
q1 = sp.Function('q1')(t)
q2 = sp.Function('q2')(t)
```

Massenmittelpunkt des Körper 1 dargestellt im Koordinatensystem {1}

```python
rc1 = sp.Matrix([lc1, 0, 0])
display(sp.Eq(sp.Symbol('rc1'), rc1, evaluate=False))
```

Massenmittelpunkt des Körper 2 dargestellt im Koordinatensystem {2}

```python
rc2 = sp.Matrix([lc2, 0, 0])
display(sp.Eq(sp.Symbol('rc2'), rc2, evaluate=False))
```

Verschiebungsvektor vom Koordinatensystem {1} ins {2}

```python
r12 = sp.Matrix([l1, 0, 0])
display(sp.Eq(sp.Symbol('r12'), r12, evaluate=False))
```

Rotationsmatrix von {1} nach {2}

```python
R12 = sp.Matrix([[sp.cos(q2), -sp.sin(q2), 0],
                 [sp.sin(q2),  sp.cos(q2), 0],
                 [0, 0, 1]])
display(sp.Eq(sp.Symbol('R12'), R12, evaluate=False))
```

## Vorwärtsrekursion

### Körper 1

Winkelgeschwindigkeit des Körpers 1

```python
omega1 = sp.Matrix([0, 0, sp.diff(q1, t)])
display(sp.Eq(sp.Symbol('omega1'), omega1, evaluate=False))
```

Winkelbeschleunigung des Körpers 1

```python
domega1 = sp.Matrix([0, 0, sp.diff(q1, (t, 2))])
display(sp.Eq(sp.Symbol('domega1'), domega1, evaluate=False))
```

Linearbeschleunigung im Ursprung des Körpers 1

```python
dv1 = sp.Matrix([0, 0, -g])
display(sp.Eq(sp.Symbol('dv1'), dv1, evaluate=False))
```

Beschleunigung im Massenmittelpunkt des Körpers 1

```python
dvc1 = sp.simplify(dv1 + domega1.cross(rc1) + omega1.cross(omega1.cross(rc1)))
display(sp.Eq(sp.Symbol('dvc1'), dvc1, evaluate=False))
```

### Körper 2

Winkelgeschwindigkeit des Körpers 2

```python
omega2 = R12.T*omega1 + sp.Matrix([0, 0, sp.diff(q2, t)])
display(sp.Eq(sp.Symbol('omega2'), omega2, evaluate=False))
```

Winkelbeschleunigung des Körpers 2

```python
domega2 = (R12.T*domega1 + sp.Matrix([0, 0, sp.diff(q2, (t, 2))])
           + omega1.cross(sp.Matrix([0, 0, sp.diff(q2, t)])))
display(sp.Eq(sp.Symbol('domega2'), domega2, evaluate=False))
```

Linearbeschleunigung im Ursprung des Körpers 2

```python
dv2 = sp.simplify(R12.T*(dv1 + domega1.cross(r12) + omega1.cross(omega1.cross(r12))))
display(sp.Eq(sp.Symbol('dv2'), dv2, evaluate=False))
```

Beschleunigung im Massenmittelpunkt des Körpers 2

```python
dvc2 = sp.simplify(dv2 + domega2.cross(rc2) + omega2.cross(omega2.cross(rc2)))
display(sp.Eq(sp.Symbol('dvc2'), dvc2, evaluate=False))
```

## Rückwärtsrekursion

### Körper 2

Resultierende Kraft im Massenmittelpunkt des Körpers 2

```python
F2 = m2*dvc2
display(sp.Eq(sp.Symbol('F2'), F2, evaluate=False))
```

Resultierendes Moment des Körpers 2

```python
N2 = I2*domega2 + omega2.cross(I2*omega2)
display(sp.Eq(sp.Symbol('N2'), N2, evaluate=False))
```

Im Gelenk 2 übertragene Kraft

```python
f2 = F2
display(sp.Eq(sp.Symbol('f2'), f2, evaluate=False))
```

Im Gelenk 2 übertragenes Moment

```python
tau2 = sp.simplify(N2 + rc2.cross(F2))
display(sp.Eq(sp.Symbol('tau2'), tau2, evaluate=False))
```

Gelenkmoment des Gelenks 2

```python
Q2 = (sp.Matrix([[0, 0, 1]])*tau2)[0]
Q2 = sp.collect(sp.expand(Q2), [sp.diff(q1, (t, 2)), sp.diff(q2, (t, 2)),
                                sp.diff(q1, t), sp.diff(q2, t)])
display(sp.Eq(sp.Symbol('Q2'), Q2, evaluate=False))
```

### Körper 1

Resultierende Kraft im Massenmittelpunkt des Körpers 1

```python
F1 = m1*dvc1
display(sp.Eq(sp.Symbol('F1'), F1, evaluate=False))
```

Resultierendes Moment des Körpers 1

```python
N1 = I1*domega1 + omega1.cross(I1*omega1)
display(sp.Eq(sp.Symbol('N1'), N1, evaluate=False))
```

Im Gelenk 1 übertragene Kraft

```python
f1 = sp.simplify(R12*f2 + F1)
display(sp.Eq(sp.Symbol('f1'), f1, evaluate=False))
```

Im Gelenk 1 übertragenes Moment

```python
tau1 = sp.simplify(N1 + R12*tau2 + rc1.cross(F1) + r12.cross(R12*F2))
display(sp.Eq(sp.Symbol('tau1'), tau1, evaluate=False))
```

Gelenkmoment des Gelenks 1

```python
Q1 = (sp.Matrix([[0, 0, 1]])*tau1)[0]
Q1 = sp.collect(sp.expand(Q1), [sp.diff(q1, (t, 2)), sp.diff(q2, (t, 2)),
                                sp.diff(q1, t), sp.diff(q2, t)])
display(sp.Eq(sp.Symbol('Q1'), Q1, evaluate=False))
```
